```python
df_clean=df.copy()
df_clean.columns
```

```python
#Handling missing values
df_clean["Age"]=df_clean["Age"].fillna(df_clean["Age"].median())
df_clean["Embarked"]=df_clean["Embarked"].fillna(df_clean["Embarked"].mode()[0])
df_clean =df_clean.drop(["Cabin","PassengerId","Name","Ticket"], axis=1)

# we are dropping name and ticket column because they dont give us any useful information about weather the person survived or not 
```

```python
from sklearn.preprocessing import LabelEncoder

le=LabelEncoder()
df_clean["Sex"]=le.fit_transform(df["Sex"])
```

```python
from sklearn.preprocessing import OneHotEncoder

on=OneHotEncoder()
encoded=on.fit_transform(df[["Embarked"]])

encoded_df=pd.DataFrame(
    encoded.toarray(),
    columns= on.get_feature_names_out(["Embarked"]),
    index=df_clean.index
)

df_clean= df_clean.drop("Embarked", axis=1)
df_clean = pd.concat([df_clean, encoded_df], axis=1)
```

```python
df_clean.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
scaled= scaler.fit_transform(df_clean[["Age","Fare"]])
df_clean[["Age","Fare"]]=scaled
```

```python
x=df_clean.drop("Survived", axis=1)
y=df_clean["Survived"]
```
